## Step 3: Data Cleaning
    • Handle missing values: Drop / Impute (mean, median, mode)
    • Remove duplicates
    • Fix incorrect data types
    • Standardize column names

```python
# Missing Values Handling 

num_cols = df.select_dtypes(include = ['int64','float64']).columns
for col in num_cols:
    df[col] = df[col].fillna(df[col].median())

cat_cols = df.select_dtypes(include = 'object').columns
for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
print(df.duplicated().sum())
# Remove Duplicates
df.drop_duplicates(inplace = True)
```

```python
# Fix incorrect data types
df['default '] = df['default '].astype('category')
```

```python
# Standardize column names
df.columns = df.columns.str.strip()
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ' , '_')
print(df.columns)
```

# Step 5: Outlier Detection & Treatment

```python
num_cols = df.select_dtypes(include=["int64","float64"]).columns

for col in num_cols:
    sns.boxplot(x=df[col])
    plt.title(f"Boxplot of {col}")
    plt.show()
```

```python
for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]
```

### Step 4: Overfitting & Underfitting Check
• Compare: Training vs Testing performance

```python
def evaluate_model(model, X_train, X_test, y_train, y_test, model_name):

    # Predictions
    y_train_pred = model.predict(X_train)
    y_test_pred  = model.predict(X_test)

    # Accuracy Scores
    train_acc = accuracy_score(y_train, y_train_pred)
    test_acc  = accuracy_score(y_test, y_test_pred)

    print("\n===================================")
    print(model_name)
    print("===================================")

    print("Training Accuracy :", round(train_acc,4))
    print("Testing Accuracy  :", round(test_acc,4))

    # Overfitting / Underfitting Check
    if train_acc - test_acc > 0.10:
        overfit = "Yes"
        remarks = "Model is Overfitting"

    elif train_acc < 0.70 and test_acc < 0.70:
        overfit = "No"
        remarks = "Model is Underfitting"

    else:
        overfit = "No"
        remarks = "Good Performance"

    print("Overfitting :", overfit)
    print("Remarks      :", remarks)

    return [model_name, train_acc, test_acc, overfit, remarks]
```

```python
### SPRINT 3
from sklearn.pipeline import Pipeline
from sklearn.feature_selection import SelectKBest
from sklearn.feature_selection import f_classif

pipelines = {

    "Logistic Regression": Pipeline([
        ("preprocessor", ct),
        ("feature_selection",
         SelectKBest(score_func=f_classif)),
        ("model",
         LogisticRegression(max_iter=1000))
    ]),

    "Decision Tree": Pipeline([
        ("preprocessor", ct),
        ("feature_selection",
         SelectKBest(score_func=f_classif)),
        ("model",
         DecisionTreeClassifier())
    ]),

    "Random Forest": Pipeline([
        ("preprocessor", ct),
        ("feature_selection",
         SelectKBest(score_func=f_classif)),
        ("model",
         RandomForestClassifier())
    ]),

    "SVM": Pipeline([
        ("preprocessor", ct),
        ("feature_selection",
         SelectKBest(score_func=f_classif)),
        ("model",
         SVC())
    ]),

    "Naive Bayes": Pipeline([
        ("preprocessor", ct),
        ("feature_selection",
         SelectKBest(score_func=f_classif)),
        ("model",
         GaussianNB())
    ])
}
```

```python
param_grid = [

    {
        "model": [LogisticRegression(max_iter=1000)],
        "feature_selection__k": [10, 15],
        "model__C": [0.1, 1, 10]
    },

    {
        "model": [DecisionTreeClassifier()],
        "feature_selection__k": [10, 15],
        "model__max_depth": [5, 10],
        "model__min_samples_split": [2, 5]
    },

    {
        "model": [RandomForestClassifier()],
        "feature_selection__k": [10, 15],
        "model__n_estimators": [100, 200],
        "model__max_depth": [5, 10]
    },

    {
        "model": [SVC()],
        "feature_selection__k": [10, 15],
        "model__C": [0.1, 1],
        "model__kernel": ["linear", "rbf"]
    },


    {
        "model": [GaussianNB()],
        "feature_selection__k": [10, 15]
    }

]
```

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python

pipe = pipelines["Random Forest"]   

random_search = RandomizedSearchCV(
    estimator=pipe,
    param_distributions=param_grid,
    n_iter=5,
    cv=3,
    scoring="accuracy",
    n_jobs=-1,
    random_state=42
)

random_search.fit(x_train, y_train)
```

```python
print("Best Parameters:", random_search.best_params_)
print("Best CV Accuracy:", random_search.best_score_)
```

```python
best_model = random_search.best_estimator_
y_pred = best_model.predict(x_test)

print("Test Accuracy :",accuracy_score(y_test, y_pred))
```

```python
from sklearn.model_selection import KFold
from sklearn.metrics import accuracy_score
import numpy as np

kf = KFold(n_splits=5, shuffle=True, random_state=42)

train_accuracies = []
test_accuracies = []

for train_index, test_index in kf.split(X):
    X_train_k, X_test_k = X.iloc[train_index], X.iloc[test_index]
    y_train_k, y_test_k = y.iloc[train_index], y.iloc[test_index]

    best_model.fit(X_train_k, y_train_k)

    y_train_pred = best_model.predict(X_train_k)
    y_test_pred  = best_model.predict(X_test_k)

    train_accuracies.append(accuracy_score(y_train_k, y_train_pred))
    test_accuracies.append(accuracy_score(y_test_k, y_test_pred))
```

```python
print("Train Accuracies:", train_accuracies)
print("Test Accuracies :", test_accuracies)

print("\nMean Train Accuracy:", np.mean(train_accuracies))
print("Mean Test Accuracy :", np.mean(test_accuracies))

print("\nTrain Accuracy Variance:", np.var(train_accuracies))
print("Test Accuracy Variance :", np.var(test_accuracies))
```
